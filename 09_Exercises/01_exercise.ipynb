2. **Stiffness.** For decay rates 1 and 100, can forward Euler use a step of 0.025? Compare backward Euler's fast-mode amplification.
3. **Stability.** On a periodic grid, take Courant number 0.4 and diffusion ratio 0.35. Does the explicit upwind/centred-diffusion update satisfy the nonnegative-weights bound?

## Worked answers

1. The time scale is 4 s; the remaining fraction is exp(-2).
2. The fast-mode forward factor is -1.5, so it grows in magnitude. Backward Euler gives 1/3.5 and decays.
3. The sum is 1.1, so this sufficient bound fails; the alternating mode also grows because its factor is 1 - 2(1.1) = -1.2.

Run the checks below, then apply the same reasoning to the tracer capstone.
