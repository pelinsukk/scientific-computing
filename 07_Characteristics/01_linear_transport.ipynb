## Model and reasoning

For f_t+v(x,t)f_x=0, choose x'(t)=v(x,t). The chain rule then gives d[f(x(t),t)]/dt=0, so each characteristic carries its initial value.

For constant v, x=x0+vt and f(x,t)=f0(x-vt). For v=x*sin(t), separation gives x=x0*exp(1-cos(t)); inversion gives x0=x*exp(-1+cos(t)). With f0(x)=1+1/(1+x²), substitute this footpoint into f0.

This is the advective equation, not the conservative equation f_t+(vf)_x=0 when v varies spatially; the latter contains an additional v_x*f term.

## Reproducible experiment

Predict the plot before running. All parameters below are illustrative, not measured environmental data.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({"figure.figsize": (8, 4), "axes.grid": True})
rng = np.random.default_rng(42)
```

```python
t=np.linspace(0,2*np.pi,300)
for x0 in np.linspace(-2,2,9):plt.plot(x0*np.exp(1-np.cos(t)),t)
plt.xlabel("position x");plt.ylabel("time");plt.title("Characteristics: dx/dt=x sin(t)");plt.show()
x=np.linspace(-10,10,400)
for ti in [0,np.pi,2*np.pi]:
 foot=x*np.exp(-1+np.cos(ti));f=1+1/(1+foot**2)
 plt.plot(x,f,label=f"t={ti:.2f}")
plt.xlabel("position");plt.ylabel("f");plt.legend();plt.show()
x0=.7;path=x0*np.exp(1-np.cos(t));carried=1+1/(1+(path*np.exp(-1+np.cos(t)))**2)
assert np.allclose(carried,1+1/(1+x0*x0))
```

## Read the result

Characteristic curves expand and contract while preserving transported values. A moving curve is not a fixed spatial observation point.
