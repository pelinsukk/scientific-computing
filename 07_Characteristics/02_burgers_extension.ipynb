## Model and reasoning

For inviscid Burgers, u_t+u*u_x=0, a smooth solution follows x=xi+t*u0(xi), u=u0(xi). The mapping derivative is 1+t*u0'(xi). If the initial slope is negative somewhere, the first loss of invertibility occurs at t_s=-1/min(u0').

Take u0(xi)=-sin(xi), giving t_s=1. Before this time the parametric curve defines a single-valued smooth solution. After crossing, extending characteristics yields multiple values at the same x and is not a physical single-valued solution. A weak entropy solution requires a shock construction or a conservative shock-capturing method, beyond the supplied notes. This experiment stops at the onset.

## Reproducible experiment

Predict the plot before running. All parameters below are illustrative, not measured environmental data.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({"figure.figsize": (8, 4), "axes.grid": True})
rng = np.random.default_rng(42)
```

```python
xi=np.linspace(-np.pi,np.pi,501);u0=-np.sin(xi)
fig,ax=plt.subplots(1,2,figsize=(10,4))
for t in [0,.4,.8,.99]:ax[0].plot(xi+t*u0,u0,label=f"t={t}")
times=np.linspace(0,1,150)
for z in np.linspace(-2,2,21):ax[1].plot(z-times*np.sin(z),times)
ax[0].set(xlabel="position",ylabel="u");ax[0].legend();ax[1].set(xlabel="position",ylabel="time");plt.tight_layout();plt.show()
assert np.min(1-.99*np.cos(xi))>0
assert np.isclose(np.min(1-np.cos(xi)),0)
print("First crossing time = 1")
```

## Read the result

The solution steepens because different values travel at different speeds. This is nonlinear deformation, unlike constant-speed linear translation.
