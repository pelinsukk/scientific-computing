## Model and reasoning

For a small-angle pendulum, $T\propto L^\alpha g^\beta m^\gamma$. Matching length, time and mass gives $\alpha+\beta=0$, $-2\beta=1$, $\gamma=0$. Thus $T=\sqrt{L/g}\,\Phi(\theta_0)$: dimensions do not determine the dimensionless function or its small-angle constant $2\pi$.

A dimension matrix contains one column per quantity; independent dimensionless groups are its nullspace. For $N$ quantities and rank $r$, there are $N-r$ independent groups (count the output among the quantities if included).

For $m u''+c u'+ku=F_0\sin(\omega t)$, set $u^*=ku/F_0$, $t^*=\omega t$. The chain rule gives $m^*u^*_{t^*t^*}+c^*u^*_{t^*}+u^*=\sin(t^*)$, with $m^*=m\omega^2/k$, $c^*=c\omega/k$. The forcing is **sin(t*)**, correcting the extra frequency printed in equation 2.23.

For decay, $u=I\bar u$, $t=\bar t/a$ gives $d\bar u/d\bar t=-\bar u$. All positive $a,I$ collapse to one curve.

## Reproducible experiment

Predict the plot before running. All parameters below are illustrative, not measured environmental data.

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({"figure.figsize": (8, 4), "axes.grid": True})
rng = np.random.default_rng(42)
```

```python
A=np.array([[1,1,0],[0,-2,0],[0,0,1]],float)
exponents=np.linalg.solve(A,[0,1,0]);assert np.allclose(exponents,[.5,-.5,0])
fig,ax=plt.subplots(1,2,figsize=(10,4))
for a,I in [(0.5,2),(1,1),(2,3)]:
 t=np.linspace(0,5/a,200);u=I*np.exp(-a*t)
 ax[0].plot(t,u,label=f"a={a}, I={I}");ax[1].plot(a*t,u/I)
ax[0].set(xlabel="time",ylabel="u");ax[0].legend();ax[1].set(xlabel="a t",ylabel="u / I");plt.tight_layout();plt.show()
print("Pendulum exponents:",exponents)
```
