# State in Computational Science

Every simulation starts with one important question:

**What is the current state of the system?**

A state contains all the information needed to describe a system at one moment in time.

If we know the current state and the mathematical model, we can predict the next state.

In this notebook, we will learn what a state is and how computers represent it.

## Learning Goals

After this notebook, you will be able to

- explain what a state is;

- identify the state of different physical systems;

- represent a state using Python;

- distinguish between a scalar, a vector, and a field;

- understand why the state is the starting point of every simulation.

## Why Is This Important?

Every simulation follows the same workflow.

Real-world system

↓

State

↓

Mathematical model

↓

Algorithm

↓

New state

Everything starts with the state.

Without a state, there is nothing to simulate.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Example 1 – Room Temperature

Suppose we want to describe a room.

If we only care about its temperature, the state is simply one number.

That number is enough to describe the system.

```python
# a scalar state

room_temperature = 20

print(room_temperature)
```

### Discussion

The system has only one variable.

The state is called a **scalar** because it contains only one value.

## Example 2 – A Moving Car

A moving car cannot be described using only one number.

We need

- position;
- velocity.

Together, they form the state.

```python
position = 10.0
velocity = 3.5

print(position)
print(velocity)
```

Instead of storing two separate variables, we can store them together.

```python
car_state = np.array([10.0, 3.5])
print(car_state)
```

### Discussion

The state now contains two values.

This is called a **state vector**.

Many engineering systems use state vectors.

## Example 3 – Temperature Along a Metal Rod

Imagine a metal rod.

The temperature is different at every point.

One number is no longer enough.

The state becomes a collection of many temperatures.

```python
temperature = np.zeros(50)

print(temperature)
```

Every element represents the temperature at one location.

The computer stores all temperatures inside one array.

```python
temperature[20] = 100

plt.figure(figsize=(8,4))

plt.plot(temperature)

plt.xlabel("position")

plt.ylabel("temperature")

plt.title("Temperature Along a Metal Rod")

plt.grid()

plt.show()
```

### Discussion

The system is no longer described by one value.

The state changes from one position to another.

This type of state is called a **field**.

## Example 4 – Weather

Weather models are much more complex.

The state may contain

- temperature;
- pressure;
- humidity;
- wind speed.

Each quantity exists at many locations.

A weather model stores millions of values inside its state.

```python
grid_size = 100

temperature = np.random.normal(20, 3, grid_size)

pressure = np.random.normal(1013, 5, grid_size)

humidity = np.random.uniform(30, 90, grid_size)
```

```python
plt.figure(figsize=(8,4))

plt.plot(temperature,label="Temperature")

plt.legend()

plt.grid()

plt.show()
```

The computer does not understand weather.

It only stores numbers.

These numbers together are called the **state**.

## Types of States

There are three common types of states.

| State Type | Example |
|------------|----------|
| Scalar | Temperature inside one room |
| Vector | Position and velocity of a car |
| Field | Temperature inside a metal rod |

## State in Python

Different mathematical objects become different Python objects.

| Mathematics | Python |
|-------------|---------|
| Scalar | float |
| Vector | NumPy array |
| Field | NumPy array |
