# Mathematical Functions

Functions are one of the most important ideas in mathematics and scientific computing.

A function takes an input and produces an output.

Many physical models are written as mathematical functions. Before solving differential equations, we need to understand how functions work and how computers evaluate them.

In this notebook, we will learn how mathematical functions are represented in Python and how to visualize them.

````python
import numpy as np
import matplotlib.pyplot as plt
````

## What Is a Function?

A function connects one input to exactly one output.

Mathematically,

$$
y=f(x)
$$

This means that the output $y$ depends on the input $x$.

Examples

$$
f(x)=x^2
$$

$$
f(x)=\sin(x)
$$

$$
f(x)=e^x
$$

Each input produces exactly one output.

````python
x = 2
y= x**2

print("Input:", x)
print("Output:", y)
````

## Evaluating a Function

Consider the function

$$
f(x)=x^2
$$

If the input is

$$
x=2,
$$

the output is

$$
f(2)=2^2=4.
$$

This process is called function evaluation.

## One Input and One Output

In the previous example, the function used only one input value.

A computer can also evaluate the same function for many input values.

This is useful because we can study the shape and behavior of a function over an interval.

````python
x_values = np.array([-3, -2, -1, 0, 1, 2, 3])

y_values = x_values**2

print("Input values:")
print(x_values)

print()

print("Output values:")
print(y_values)
````

## Evaluating Many Inputs

For the function

$$
f(x)=x^2,
$$

the computer applies the same rule to every value in the array.

For example,

$$
f(-3)=9,
$$

$$
f(-2)=4,
$$

$$
f(-1)=1,
$$

and

$$
f(0)=0.
$$

NumPy performs these calculations automatically.

## Domain

The domain is the set of all valid input values of a function.

For the function

$$
f(x)=x^2,
$$

every real number can be used as an input.

In this notebook, we will study the function on the interval

$$
-3 \leq x \leq 3.
$$

A computer cannot use infinitely many points. Therefore, we represent the interval with a finite set of values.

````python
x_values = np.linspace(-3.0, 3.0, 200)

print("Number of points:", len(x_values))
print("First five values:", x_values[:5])
print("Last five values:", x_values[-5:])
````

## A Discrete Grid

The command 'np.linspace() creates equally spaced points.
In this example, the continuous interval

$$
[-3,3]
$$

is represented by 200 points.

This is a simple example of discritization.

Discritization means replacing a continuous domain with a finite set of points that a computer can use.

# Range 

The range is the set of output values produced by a function.

For 

$$
f(x)=x^2,
$$

the output can never be negative.

On the interval

$$
-3 \leq x \leq 3,
$$

the smallest output is

$$
0,
$$

and the largest output is

$$
9,
$$

````python
y_values = x_values**2

print("Minimum output:", y_values.min())
print("Maximum:", y_values.max())
````

## Plotting a Function

A graph shows how the output changes when the input changes.

The horizontal axis represents the input values.

The vertical axis represents the output values.

We will now plot

$$
f(x)=x^2\dots
$$

````python
plt.figure(figsize=(8, 5))

plt.plot(
    x_values,
    y_values,
    label="f(x) = x^2",
    linewidth=2
)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Quadratic Function")
plt.grid()
plt.legend()
plt.show()
````

## Interpreting the Graph

The graph has a U-shape.

The function reaches its minimum value at

$$
x=0.
$$

The graph is symmetric around the vertical axis.

This means that

$$
f(-x)=f(x).
$$

For example,

$$
f(-2)=f(2)=4.
$$

## Functions in Python

We can define our own functions in Python.

A Python function receives an input, applies a rule, and returns an output.

The following function represents

$$
f(x)=x^2.
$$

````python
def square_function(x):
    return x**2
````

````python
result = square_function(5)

print("Input:", 5)
print("Output:", result)
````

## Understanding the Python Function

The line


```python
def square_function(x):

creates a function called square_function.

The variable x is the input.

The line

return x**2

calculates the output and returns it.

````python
test_values = np.array([-2, -1, 0, 1, 2])

results = square_function(test_values)

print(results)
````

## Functions and NumPy Arrays

The same Python function can receive a NumPy array.

The operation

x^2

is applied to every element in the array.

This is called vectorized computation.

Vectorized operations are important because they are simple, clear and fast.

## Common Scientific Functions

NumPy provides many mathematical functions.

Some common examples are

$$
\sin(x),
$$

$$
\cos(x),
$$

$$
e^x,
$$

$$
\sqrt{x},
$$

and

$$
\ln(x).
$$

These functions are widely used in scientific models.

````python
x_trig = np.linspace(-2.0 * np.pi, 2.0 * np.pi, 400)

sine_values = np.sin(x_trig)
cosine_values = np.cos(x_trig)
````

````python
plt.figure(figsize=(10, 5))

plt.plot(
    x_trig,
    sine_values,
    label="sin(x)",
    linewidth=2
)

plt.plot(
    x_trig,
    cosine_values,
    label="cos(x)",
    linewidth=2
)

plt.xlabel("x")
plt.ylabel("Function value")
plt.title("Sine and Cosine Functions")
plt.grid()
plt.legend()
plt.show()
````

## Sine and Cosine

The sine and cosine functions are periodic.

This means that their values repeat after a fixed interval.

Both functions have values between

$$
-1
$$

and

$$
1.
$$

These functions are used to describe waves, oscillations, and rotating systems.

## The Exponential Function

The exponential function is written as

$$
f(x)=e^x.
$$

It grows slowly for negative values of $x$ and very quickly for positive values of $x$.

Exponential functions are used in growth, decay, population models, heat transfer, and differential equations.

````python
x_exp = np.linspace(-3.0, 3.0, 300)
exponential_values = np.exp(x_exp)
````

````python
plt.figure(figsize=(8, 5))

plt.plot(
    x_exp,
    exponential_values,
    label="exp(x)",
    linewidth=2
)

plt.xlabel("x")
plt.ylabel("exp(x)")
plt.title("Exponential Function")
plt.grid()
plt.legend()
plt.show()
````

## The Logarithm Function

The natural logarithm is written as

$$
f(x)=\ln(x).
$$

The logarithm is only defined for positive input values.

Therefore, its domain is

$$
x>0.
$$

Using zero or a negative value produces an invalid mathematical result.

````python
x_log = np.linspace(0.1, 5.0, 300)

logarithm_values = np.log(x_log)
````

````python
plt.figure(figsize=(8, 5))

plt.plot(
    x_log,
    logarithm_values,
    label="ln(x)",
    linewidth=2
)

plt.xlabel("x")
plt.ylabel("ln(x)")
plt.title("Natural Logarithm")
plt.grid()
plt.legend()
plt.show()
````

## The Square Root Function

The square root function is

$$
f(x)=\sqrt{x}.
$$

For real-valued calculations, the input must satisfy

$$
x\geq0.
$$

The function grows as $x$ increases, but its growth becomes slower.

````python
x_sqrt = np.linspace(0.0, 100, 300)

square_root_values = np.sqrt(x_sqrt)
````

````python
plt.figure(figsize=(8, 5))

plt.plot(
    x_sqrt,
    square_root_values,
    label="sqrt(x)",
    linewidth=2
)

plt.xlabel("x")
plt.ylabel("sqrt(x)")
plt.title("Square Root Function")
plt.grid()
plt.legend()
plt.show()
````

## Comparing Different Functions

Different functions can grow at different rates.

We will compare

$$
x,
$$

$$
x^2,
$$

$$
x^3,
$$

and

$$
e^x.
$$

This comparison helps us understand the behavior of mathematical models.

````python
x_compare = np.linspace(0.0, 3.0, 300)

linear_values = x_compare
quadratic_values = x_compare**2
cubic_values = x_compare**3
exponential_values = np.exp(x_compare)
````

````python
plt.figure(figsize=(10, 6))

plt.plot(
    x_compare,
    linear_values,
    label="x"
)

plt.plot(
    x_compare,
    quadratic_values,
    label="x²"
)

plt.plot(
    x_compare,
    cubic_values,
    label="x³"
)

plt.plot(
    x_compare,
    exponential_values,
    label="exp(x)"
)

plt.xlabel("x")
plt.ylabel("Function value")
plt.title("Comparison of Function Growth")
plt.grid()
plt.legend()
plt.show()
````

## Discussion

The linear function grows at a constant rate.

The quadratic and cubic functions grow faster as $x$ increases.

The exponential function also grows rapidly.

The function that appears largest depends on the selected interval. Therefore, graphs should always be interpreted together with their domain.
